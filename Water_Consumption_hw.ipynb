```python
# Water Consumption Calculator
people = float(input("Enter the amount of people in your household: "))
daily_usage = float(input("Enter the average daily water usage per person in liters: "))
price_per_liter = float(input("Enter the price per liter: "))

# Calculate total monthly consumption
monthly_consumption = people * daily_usage * 30

# Calculate total cost
total_cost = monthly_consumption * price_per_liter

# Display results
print("Total monthly consumption:", monthly_consumption, "liters")
print("Total monthly cost:", total_cost)
```
